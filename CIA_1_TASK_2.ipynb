# Step 2: Import libraries


```python
from transformers import T5Tokenizer, T5ForConditionalGeneration
import torch
```

# Step 3: Load model and tokenizer


```python
model_name = "google/flan-t5-large"
tokenizer = T5Tokenizer.from_pretrained(model_name)
model = T5ForConditionalGeneration.from_pretrained(model_name)
```

# Step 4: Define prompt list


```python
prompts = [
    "The Lost City of Code",
    "A Journey Through the AI Forest",
    "The Day Robots Took Over"
]
```

# Step 5: Define the generation function

```python
def generate_story(prompt):
    input_text = f"Write a short story about: {prompt}"
    input_ids = tokenizer.encode(input_text, return_tensors="pt", truncation=True)

    output_ids = model.generate(
        input_ids=input_ids,
        max_length=256,
        temperature=0.9,
        top_k=50,
        top_p=0.95,
        do_sample=True,
        num_beams=2,
        early_stopping=True
    )

    return tokenizer.decode(output_ids[0], skip_special_tokens=True)
```

# Step 6: Generate stories


```python
# ✅ Step 6: Generate stories
for prompt in prompts:
    print("=" * 50)
    print(f"🧵 Prompt: {prompt}")
    story = generate_story(prompt)
    print(f"📖 Story:\n{story}\n")
```
